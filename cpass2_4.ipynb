```python
def solve_n_queens(n):
    # Initialize an empty board
    board = [['.' for _ in range(n)] for _ in range(n)]
    solutions = []

    def is_safe(row, col):
        # Check this row on left side
        for i in range(col):
            if board[row][i] == 'Q':
                return False

        # Check upper diagonal on left side
        for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
            if board[i][j] == 'Q':
                return False

        # Check lower diagonal on left side
        for i, j in zip(range(row, n, 1), range(col, -1, -1)):
            if board[i][j] == 'Q':
                return False

        return True

    def backtrack(col):
        # Base case: If all queens are placed, add the current board to solutions
        if col >= n:
            current_solution = ["".join(row) for row in board]
            solutions.append(current_solution)
            return

        # Recursive case: Try placing a queen in the current column
        for i in range(n):
            if is_safe(i, col):
                board[i][col] = 'Q'  # Place queen
                backtrack(col + 1)   # Recur for the next column
                board[i][col] = '.'  # Backtrack: Remove queen (explore other possibilities)

    backtrack(0)
    return solutions

def print_solutions(solutions):
    if not solutions:
        print("No solutions found.")
        return

    print(f"Found {len(solutions)} solutions for N-Queens.")
    for i, solution in enumerate(solutions):
        print(f"\nSolution {i+1}:")
        for row_str in solution:
            print(row_str)

# Example usage for N = 4
N = int(input("Enter the value of N for N-Queens: "))
print(f"Solving N-Queens for N = {N}:")
n_queens_solutions = solve_n_queens(N)
print_solutions(n_queens_solutions)

```

```python
def permute_string(s):
    chars = list(s)
    n = len(chars)
    results = []

    def backtrack(index):
        if index == n:
            results.append("".join(chars))
            return

        for i in range(index, n):
            chars[index], chars[i] = chars[i], chars[index]  # Swap characters
            backtrack(index + 1)                             # Recurse for the next index
            chars[index], chars[i] = chars[i], chars[index]  # Backtrack: Undo the swap

    backtrack(0)
    return results

# Sample Input
input_string = input("Enter a string to permute: ")

print(f"\nPermutations for '{input_string}':")
permutations = permute_string(input_string)
for p in permutations:
    print(p)

print(f"\nTotal permutations found: {len(permutations)}")
```

```python
def subset_sum(expenses, target_budget):
    n = len(expenses)

    def backtrack(index, current_sum):
        # Base Case 1: If current_sum exactly matches target_budget, we found a solution
        if current_sum == target_budget:
            return True

        # Base Case 2: If current_sum exceeds target_budget OR we've exhausted all expenses
        if current_sum > target_budget or index == n:
            return False

        # Recursive Case: Try two paths for the current expense at 'index'

        # Path 1: Include the current expense
        if backtrack(index + 1, current_sum + expenses[index]):
            return True

        # Path 2: Exclude the current expense
        if backtrack(index + 1, current_sum):
            return True

        return False # No solution found from this path

    return backtrack(0, 0)

# Sample Input
num_expenses = int(input("Enter the number of expenses: "))
expenses_list_str = input(f"Enter {num_expenses} expenses separated by spaces: ")
expenses = list(map(int, expenses_list_str.split()))
target_budget = int(input("Enter the target budget: "))

print(f"\nExpenses: {expenses}")
print(f"Target Budget: {target_budget}")

if subset_sum(expenses, target_budget):
    print("Output: True (A combination of expenses exactly matches the target budget)")
else:
    print("Output: False (No combination of expenses exactly matches the target budget)")
```

```python
def solve_sudoku(board):
    # Function to find the next empty cell (marked as 0)
    def find_empty(board):
        for r in range(9):
            for c in range(9):
                if board[r][c] == 0:
                    return (r, c)  # Return (row, col) of the empty cell
        return None  # No empty cells, board is full

    # Function to check if placing 'num' at (row, col) is valid
    def is_valid(board, num, pos):
        row, col = pos

        # Check row
        for c in range(9):
            if board[row][c] == num and col != c:
                return False

        # Check column
        for r in range(9):
            if board[r][col] == num and row != r:
                return False

        # Check 3x3 box
        box_x = col // 3
        box_y = row // 3

        for r_box in range(box_y * 3, box_y * 3 + 3):
            for c_box in range(box_x * 3, box_x * 3 + 3):
                if board[r_box][c_box] == num and (r_box, c_box) != pos:
                    return False
        return True

    # Main backtracking function
    def backtrack():
        find = find_empty(board)
        if not find:
            return True  # No empty cells left, puzzle solved!
        else:
            row, col = find

        for num in range(1, 10):
            if is_valid(board, num, (row, col)):
                board[row][col] = num  # Place the number

                if backtrack():  # Recurse with the new board state
                    return True

                board[row][col] = 0  # Backtrack: Reset the cell if recursion didn't find a solution
        return False  # No valid number could be placed in this cell

    # Start the backtracking process
    if backtrack():
        return board
    else:
        return None # No solution exists

def print_board(board):
    if board is None:
        print("No solution exists for this Sudoku puzzle.")
        return
    for r in range(len(board)):
        if r % 3 == 0 and r != 0:
            print("- - - - - - - - - - - -")

        for c in range(len(board[0])):
            if c % 3 == 0 and c != 0:
                print(" | ", end="")

            if c == 8:
                print(board[r][c])
            else:
                print(str(board[r][c]) + " ", end="")

# Sample Input (Partial Grid)
sample_board = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]

print("Original Sudoku Board:")
print_board(sample_board)

solved_board = solve_sudoku(sample_board)

print("\nSolved Sudoku Board:")
print_board(solved_board)
```

```python
def solve_maze(maze):
    n = len(maze)
    # Create a solution matrix, initialized with 0s
    solution = [[0 for _ in range(n)] for _ in range(n)]

    def is_safe(x, y):
        # Check if (x, y) is within maze boundaries and is an open path (1)
        return 0 <= x < n and 0 <= y < n and maze[x][y] == 1

    def backtrack(x, y):
        # Base Case: If the destination is reached
        if x == n - 1 and y == n - 1:
            solution[x][y] = 1 # Mark the destination as part of the path
            return True

        # Check if the current position is safe to move to
        if is_safe(x, y):
            solution[x][y] = 1 # Mark current cell as part of the solution path

            # Try moving down
            if backtrack(x + 1, y):
                return True

            # Try moving right
            if backtrack(x, y + 1):
                return True

            # If neither down nor right leads to a solution, backtrack
            # Unmark the current cell as part of the solution path
            solution[x][y] = 0
            return False

        return False # If not safe, return False

    # Start backtracking from (0,0)
    if backtrack(0, 0):
        return solution
    else:
        return None # No path found

def print_solution(solution):
    if solution is None:
        print("No path exists.")
        return
    for row in solution:
        print(" ".join(map(str, row)))

# Sample Input
N = int(input("Enter the size of the maze (N): "))
print(f"Enter the {N}x{N} maze row by row, with 1 for open path and 0 for blocked cell:")
maze_input = []
for i in range(N):
    row_str = input(f"Row {i+1}: ")
    row = list(map(int, row_str.split()))
    if len(row) != N:
        print(f"Error: Row {i+1} must have {N} elements. Please re-enter.")
        exit()
    maze_input.append(row)

print("\nOriginal Maze:")
for row in maze_input:
    print(" ".join(map(str, row)))

solved_path = solve_maze(maze_input)

print("\nSolution Path (1s mark the path, 0s are unvisited/blocked):")
print_solution(solved_path)
```
